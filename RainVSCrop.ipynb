```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

data = {
    'Year': [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019],
    'Temperature (°C)': [22.5, 23.0, 21.5, 22.0, 23.5, 24.0, 22.8, 23.3, 21.8, 23.7],
    'Rainfall (mm)': [100, 110, 90, 95, 105, 115, 98, 108, 92, 112],
    'Humidity (%)': [75, 70, 80, 78, 72, 68, 74, 71, 77, 69],
    'Sunshine Hours': [2000, 1800, 2100, 1900, 1700, 1600, 1950, 1750, 2050, 1650],
    'Crop Yield (tons/acre)': [3.5, 3.6, 3.4, 3.5, 3.7, 3.8, 3.6, 3.7, 3.5, 3.8],
    'Fertilizer Usage (kg/acre)': [50, 55, 45, 52, 57, 60, 53, 56, 46, 58],
    'Pesticide Usage (liters/acre)': [10, 11, 9, 10.5, 12, 13, 10.8, 11.5, 9.5, 12.5]
}

data_set=pd.DataFrame(data)
target = "Crop Yield (tons/acre)"
x = data_set.drop(columns=[target])
y = data_set[target]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.3,random_state=42)
lr=LinearRegression()
lr.fit(x_train,y_train)
prediction=lr.predict([[2024,25,123,85,1800,60,56]])
print(f"{prediction[0]}")
```
